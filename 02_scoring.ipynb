```python
import pandas as pd
import joblib
```

```python
# Load prepared scoring data
score_al = pd.read_csv('data/01_score_al.csv')
score_apd = pd.read_csv('data/01_score_apd.csv')
```

```python
feat_cols_al = ["vehicle_count", "num_heavy_vehicles", "risk_score_external", "prior_year_mileage_000", "prior_al_claim_count",
                "driver_count", "vehicle_avg_age", "late_payment_count", "coverage_limit_000", "prior_loss_amount", 
                "business_type", "payment_frequency", "state", "deductible"]

feat_cols_apd = ["vehicle_count", "num_heavy_vehicles", "risk_score_external", "prior_year_mileage_000", "prior_apd_claim_count",
                "driver_count", "vehicle_avg_age", "late_payment_count", "coverage_limit_000", "prior_loss_amount", "annual_premium",
                "years_in_business",
                "business_type", "payment_frequency", "state", "deductible"]
```

```python
X_al = score_al[feat_cols_al]
X_apd = score_apd[feat_cols_apd]
```

```python
# Load the best models for both AL and APD
al_model = joblib.load('models/best_model_al.joblib')
apd_model = joblib.load('models/best_model_apd.joblib')
```

```python
# Predict probabilities for both AL and APD
y_al = al_model.predict_proba(X_al)[:, 1]
y_apd = apd_model.predict_proba(X_apd)[:, 1]

score_al['risk_score'] = y_al
score_apd['risk_score'] = y_apd
```

```python
# Combine the predictions into a single table
predicted = pd.concat([score_al, score_apd], axis=0).sort_index()
```

```python
predicted.to_csv('data/predictions.csv', index=False)
```
